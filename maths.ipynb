```python
def count_digits(n):
    count = 0
    while(n > 0):
        last_digit = n % 10
        count = count + 1
        n = n // 10
    return count

count_digits(1324)
```

```python
def reverse_number(n):
    revN = 0
    while(n > 0):
        last_digit = n % 10
        n = n // 10
        revN = revN * 10  +  last_digit
    return revN

reverse_number(42387)
```

```python
def check_palindrome(n):
    duplicate = n
    rev = 0
    while(n > 0):
        last_D = n % 10
        n = n // 10
        rev = rev * 10 + last_D

    return duplicate == rev

check_palindrome(234234)
```

```python
def armstrong(n):
    dup = n
    sum = 0
    while(n > 0):
        last_digit = n % 10
        n = n // 10
        sum += (last_digit**3)
    return dup == sum

armstrong(153)
```

```python
def print_divisors(n):
    for i in range(1,n+1):
        if n % i == 0:
            print(i)

print_divisors(36)
```

```python
def prime(n):
    count = 0
    for i in range(1,n+1):
        if n % i == 0:
            count += 1
    return count == 2

prime(12)
```

```python
def prime_1(n):
    count = 0
    for i in range(1,int(math.sqrt(n))+1):
        if n % i == 0:
            count += 1
            
            if n/i != 1:
                count += 1

    return count == 2
prime_1(11)
```

```python
def gcd(n1,n2):
    hcf = 1
    for i in range(1,min(n1,n2)+1):
        if(n1 % i == 0 and n2 % i == 0):
            hcf = i
    return hcf

gcd(20,40)
```

```python
def gcd_1(n1,n2):
    for i in range(min(n1,n2),1,-1):
        if(n1 % i == 0 and n2 % i == 0):
            print(i)
            break
    
gcd_1(20,40)
```
