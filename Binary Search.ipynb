```python
# floor 

def floor(arr,x):
    low = 0
    high = len(arr) - 1
    ans = -1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] <= x:
            ans = arr[mid]
            low = mid + 1
        else:
            high = mid - 1
    return ans

nums = [10,20,30,40,50]
floor(nums,25)
```

```python
# find the first and last occurance index of x

def occur(arr,x):
    first = -1
    last = -1
    for i in range(len(arr)):
        if arr[i] == x:
            if first == -1:
                first = i
            last = i
    return [first,last]

nums = [2,4,6,8,8,8,11,13]

print(occur(nums,8),occur(nums,10),occur(nums,11))
```

```python
#find the sqaure root of a number in logn

n = 46

low = 1
high = n
ans = 1
while low <= high:
    mid = (low + high) // 2
    if mid*mid <= n:
        ans = max(mid,ans)

    if mid*mid > n:
        high = mid - 1
    else:
        low = mid + 1

print(ans)
```
